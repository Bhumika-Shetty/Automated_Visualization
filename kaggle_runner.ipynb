```python
# Step 3: Load API keys from Kaggle Secrets
import os
from kaggle_secrets import UserSecretsClient
secrets = UserSecretsClient()

os.environ['OPENAI_API_KEY']    = secrets.get_secret('OPENAI_API_KEY')
os.environ['ANTHROPIC_API_KEY'] = secrets.get_secret('ANTHROPIC_API_KEY')
os.environ['GOOGLE_API_KEY']    = secrets.get_secret('GOOGLE_API_KEY')
os.environ['GROQ_API_KEY']      = secrets.get_secret('GROQ_API_KEY')
os.environ['DEEPSEEK_API_KEY']  = secrets.get_secret('DEEPSEEK_API_KEY')

print('API keys loaded ✓')
```

```python
# Step 4: Configure which models to run in this session
# Change this list to run only specific models
MODELS_TO_RUN = [
    'deepseek-chat',
    'qwen-qwq-32b',
    'llama-3.1-8b-instant',
]

DATASETS_TO_RUN = [
    'titanic', 'iris', 'sleep_health', 'customer_segments', 'credit_fraud',
    'dogs_zurich', 'cameras', 'coffee_sales'
]

NUM_RUNS = 30
print(f'Will run: {len(MODELS_TO_RUN)} models × {len(DATASETS_TO_RUN)} datasets × 2 prompts × {NUM_RUNS} runs = {len(MODELS_TO_RUN)*len(DATASETS_TO_RUN)*2*NUM_RUNS} total')
```

```python
# Step 5: Run the pipeline
from pipeline_v3 import Pipeline

p = Pipeline(
    models=MODELS_TO_RUN,
    datasets=DATASETS_TO_RUN,
    num_runs=NUM_RUNS
)
p.run()
```
